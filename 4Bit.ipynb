```python
# 7B parameter
num_params = 7_000_000_000



bf16_bits_per_param = 16

bf16_total_bits = num_params * bf16_bits_per_param
bf16_total_bytes = bf16_total_bits / 8
bf16_total_gb = bf16_total_bytes / 1e9

print("BF16 weight memory:", bf16_total_gb, "GB")



quant_bits_per_param = 4

quant_total_bits = num_params * quant_bits_per_param
quant_total_bytes = quant_total_bits / 8
quant_total_gb = quant_total_bytes / 1e9

print("4-bit weight memory:", quant_total_gb, "GB")



block_size = 64

num_blocks = num_params / block_size

print("Number of blocks:", num_blocks)



scale_bits = 32

scale_total_bits = num_blocks * scale_bits
scale_total_bytes = scale_total_bits / 8
scale_total_mb = scale_total_bytes / 1e6

print("Scale memory before double quantization:",
      scale_total_mb, "MB")




scale_overhead_per_param = scale_bits / block_size

print("Scale overhead:",
      scale_overhead_per_param,
      "bits/parameter")

effective_bits = (
    quant_bits_per_param
    + scale_overhead_per_param
)

print("Effective storage before DQ:",
      effective_bits,
      "bits/parameter")




dq_scale_overhead = 0.127

dq_total_scale_bits = num_params * dq_scale_overhead
dq_total_scale_bytes = dq_total_scale_bits / 8
dq_total_scale_mb = dq_total_scale_bytes / 1e6

print("Scale memory after double quantization:",
      dq_total_scale_mb,
      "MB")




saved_bits_per_param = (
    scale_overhead_per_param
    - dq_scale_overhead
)

saved_total_bits = (
    num_params * saved_bits_per_param
)

saved_total_bytes = saved_total_bits / 8
saved_total_mb = saved_total_bytes / 1e6

print("Memory saved by double quantization:",
      saved_total_mb,
      "MB")




before_dq_gb = (
    quant_total_bytes
    + scale_total_bytes
) / 1e9

after_dq_gb = (
    quant_total_bytes
    + dq_total_scale_bytes
) / 1e9

print("Before double quantization:",
      before_dq_gb, "GB")

print("After double quantization:",
      after_dq_gb, "GB")
```
