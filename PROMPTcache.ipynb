```python
from openai import OpenAI


client = OpenAI(api_key=OPENAI_API_KEY)

shared_prompt = """
You are a coding assistant.

Here is a very long document that is reused across requests:
[imagine thousands of tokens here...]
"""

response1= client.responses.create(
    model="gpt-5",
    input=shared_prompt + "\nExplain Python decorators."
)

print(response1.usage)
```

```python
from openai import OpenAI
import time

client = OpenAI(api_key=OPENAI_API_KEY)

shared_context = """
You are a senior Python tutor.

Rules:
- Explain concepts clearly.
- Give small examples.
- Avoid unnecessary jargon.
- Explain the reasoning behind the code.

Python is a high-level programming language used for backend systems,
data science, machine learning, automation, APIs, scripting, and more.

""" * 100


def ask(question):
    start = time.perf_counter()

    response = client.responses.create(
        model="gpt-5.6",
        prompt_cache_key="python-tutor-demo",

        input=shared_context + f"""

User question:
{question}
""",

        max_output_tokens=50,
    )

    latency = time.perf_counter() - start


    print("Question:", question)
    print(f"Latency: {latency:.2f}s")

    print("Input tokens:",
          response.usage.input_tokens)

    print("Cache write tokens:",
          response.usage.input_tokens_details.cache_write_tokens)

    print("Cached tokens:",
          response.usage.input_tokens_details.cached_tokens)


ask("What is a Python decorator?")
ask("What is a Python generator?")
ask("What is a Python context manager?")
```
